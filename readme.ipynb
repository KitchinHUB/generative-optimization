# Generative Machine Learning Approaches to Optimization

**Authors:** Victor Alves and John R. Kitchin

This repository contains code examples demonstrating generative machine learning approaches to optimization problems, including:

- **Gaussian Mixture Models (GMM)** for conditional generation
- **Conditional Flow Matching** for inverse design

Both approaches learn joint distributions over inputs and outputs, enabling bidirectional inference.

## Quick Start Example

Here's a minimal example of using flow matching for inverse design:

````python
import numpy as np
from generative_optimization import generate_samples, ConditionalFlowMatching

# Define a simple forward model: y = x^2
def forward_model(x):
    return x ** 2

# Generate training data
x_data = generate_samples(bounds=[[-2, 2]], n_samples=512)
y_data = forward_model(x_data)

# Train flow matching for inverse: given y, find x
fm = ConditionalFlowMatching(x_dim=1, c_dim=1, hidden_dim=64, n_layers=3)
fm.fit(x_data, y_data, epochs=500, verbose=True)

# Inverse problem: find x where y = 1
samples = fm.sample(c_values=[[1.0]], n_samples=500)

print(f"\nTarget: y = 1.0")
print(f"Found x values: {samples.mean():.3f} ± {samples.std():.3f}")
print(f"Expected: x = ±1.0")
````
